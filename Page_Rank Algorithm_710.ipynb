```python
import numpy as np

class PageRank:

    def __init__(self, damping=0.85, iterations=100):
        self.damping = damping
        self.iterations = iterations

    def compute(self, matrix):

        n = len(matrix)

        # Convert adjacency matrix into transition matrix
        for i in range(n):
            row_sum = matrix[i].sum()

            if row_sum != 0:
                matrix[i] = matrix[i] / row_sum

        # Initial PageRank
        rank = np.ones(n) / n

        # Power iteration
        for iteration in range(self.iterations):

            new_rank = (
                (1 - self.damping) / n +
                self.damping * matrix.T @ rank
            )

            # Check if PageRank values have converged
            if np.allclose(rank, new_rank, atol=1e-6):
                print(f"\nConverged after {iteration + 1} iterations")
                rank = new_rank
                break

            rank = new_rank

        else:
            print(f"\nDid not converge after {self.iterations} iterations")

        return rank


# Input number of nodes
N = int(input("Enter number of nodes: "))

# Input adjacency matrix
matrix = []

print("Enter 1 if there is a link, otherwise enter 0")

for i in range(N):
    row = []

    for j in range(N):
        x = int(input(f"Is there a link from {i} to {j}? "))
        row.append(x)

    matrix.append(row)

# Convert to NumPy array
matrix = np.array(matrix, dtype=float)

# Compute PageRank
pr = PageRank(iterations=100)
scores = pr.compute(matrix)

# Display results
print("\nPageRank Scores:")

for i in range(N):
    print(f"Node {i}: {scores[i]:.4f}")
```
